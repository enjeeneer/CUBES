```python
import pandas as pd
import numpy as np
import datetime


years = np.arange(2017,2023)
for year in years:
    datetime_end = datetime.datetime(year,12,31,hour=23,minute=50,tzinfo=datetime.timezone.utc)
    grid_data = pd.read_csv(f"grid_carbon_GB_{year}.csv",index_col=0)
    grid_data.index=pd.DatetimeIndex(pd.to_datetime(grid_data.index,utc=True))
    grid_data.loc[datetime_end]=grid_data.loc[grid_data.index[-1]][grid_data.columns[0]]
    grid_data_upsampled=grid_data.resample("10MIN").asfreq()
    grid_data_upsampled_interpolated = grid_data_upsampled.interpolate()
    grid_data_upsampled_interpolated.to_csv(f"grid_carbon_GB_10min_{year}.csv")
```
