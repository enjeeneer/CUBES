```python
import pandas as pd
import numpy as np

xls = pd.ExcelFile('occupancy_data.xlsm')
df_init = pd.read_excel(xls, 'init_states')
df_transiton = pd.read_excel(xls, 'transition_matrix')
```

```python
weekday_df_init_state = df_init[df_init["Weekday/Weekend"] == "Weekday"].drop("Weekday/Weekend", axis=1).rename({"Number of Occupants": "number_of_occupants"}, axis=1)
weekend_df_init_state = df_init[df_init["Weekday/Weekend"] == "Weekend"].drop("Weekday/Weekend", axis=1).rename({"Number of Occupants": "number_of_occupants"}, axis=1)

weekday_df_transition = df_transiton[df_transiton["Weekday/Weekend"] == "Weekday"].drop("Weekday/Weekend", axis=1)
weekend_df_transition = df_transiton[df_transiton["Weekday/Weekend"] == "Weekend"].drop("Weekday/Weekend", axis=1)

with open("weekday_occupancy_init_states.parquet", "wb") as f:
    weekday_df_init_state.to_parquet(f)

with open("weekend_occupancy_init_states.parquet", "wb") as f:
    weekend_df_init_state.to_parquet(f)

with open("weekday_occupancy_transition.parquet", "wb") as f:
    weekday_df_transition.to_parquet(f)

with open("weekend_occupancy_transition.parquet", "wb") as f:
    weekend_df_transition.to_parquet(f)
```
