# Basic example

This is a minimal example to check elhogym functionality.
Elhogym is built on sinergym, but lets us use buildings, variables, rewards etc. which are not defined in sinergym without touching the sinergym source code.
This example tests the use of input files located in the case directory (directory names important)

```python
import gym
import numpy as np
import sys

#add elhogym to the PYTHONPATH
sys.path.insert(0, '/workspaces/elizabeth-homes/exp/hannes')
#load environment definition from input folder
import input
#make environment
env = gym.make('Eplus-1storeytest-v2')
```


Run for 1 episode.

```python
for i in range(1):
    obs = env.reset()
    rewards = []
    done = False
    current_month = 0
    while not done:
        a = env.action_space.sample()
        obs, reward, done, info = env.step(a)
        rewards.append(reward)
        if info['month'] != current_month:  # display results every month
            current_month = info['month']
            print('Reward: ', sum(rewards), info)
```

close the environment:

```python
env.close()
```

Now we can see the final rewards:

```python
print(
    'Mean reward: ',
    np.mean(rewards),
    'Cumulative reward: ',
    sum(rewards))
```
