```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

singh_data = pd.read_csv("validation_data/EPSimulation.csv")
cubes_data = pd.read_csv("output/eplusout.csv")
```

```python
time = np.arange(8760)
scols = singh_data.columns
singh_energy = np.cumsum(singh_data[scols[1]]+singh_data[scols[2]]+singh_data[scols[3]])*0
for i in range(4,9):
    singh_energy += np.cumsum(singh_data[scols[i]])
```

```python
ccols = cubes_data.columns
cubes_energy = np.cumsum(cubes_data[ccols[8]])*3600
len(cubes_data)
```

```python
fig,ax = plt.subplots()
ax.plot(time,singh_energy*2.77778e-10)
ax.plot(time,cubes_energy*2.77778e-10)
```
