```python
import pandas as pd
import numpy as np
 
gas_column = 'Site:Environmental Impact NaturalGas Source Energy [J](Hourly)'
elec_column = 'Whole Building:Facility Net Purchased Electricity Rate [W](Hourly)'
co2_column = 'Site:Environmental Impact Total CO2 Emissions Carbon Equivalent Mass [kg](Hourly) '

installation_costs = [0,3800,16000,51000,71000,23000,26800,39000,74000,94000,33000,36800,49000,84000,104000]
cases=15
reps=5
years = np.arange(2017,2023)
dir_name = "evaluation_uncontrolled"
```

```python
gas_price_per_kwh = 0.1
elec_price_per_kwh = 0.35
elec_export_price_per_kwh = 0.15

results = []
for c in range(cases):
    for y in years:
        for r in range(reps):
            data = pd.read_csv(dir_name+"/case_"+str(c)+"/year_"
                               +str(y)+"/rep_"+str(r)+"/output/eplusout.csv")
            
            co2_sum = data[co2_column].sum()/1000
            gas_sum = data[gas_column].sum()*2.777778e-7
            elec_sum = data[elec_column].sum()/1000

            if elec_sum > 0:
                bill_sum = gas_sum*gas_price_per_kwh+ elec_sum*elec_price_per_kwh
            else: 
                bill_sum = gas_sum*gas_price_per_kwh + elec_sum*elec_export_price_per_kwh

            row_dict = {"case":c,"year":y,"rep":r,
                        "gas [kWh]":gas_sum,"electricity [kWh]":elec_sum,
                        "bills [GBP]":bill_sum,"CO2 [t]":co2_sum,
                        "costs [GBP]":installation_costs[c]}
            
            results.append(row_dict)

results_df = pd.DataFrame(results)
```

```python
results_df.to_csv("results_uncontrolled.csv")
results_df
```
