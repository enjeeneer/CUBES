```python
import os
import numpy as np
import json
from input_file_factory import get_input_file_with_schedules_etc
cwd_path = os.getcwd()
print(cwd_path)
```

```python
cases = np.arange(15)
years = np.arange(2017,2023)
reps_per_year = 10
run_name = "evaluation_new"
```

```python
for case in cases:
    input_file_path = "../00_base_input/case" + str(case) + ".json"
    for y in years:
        for r in range(reps_per_year):
            case_path = cwd_path + "/" + run_name + "/case_" + str(case)+ "/year_" + str(y) + "/rep_" + str(r)
            complete_input_file_path_uc = case_path + "/input_uc.json"
            complete_input_file_path_c = case_path + "/input_c.json"

            if not os.path.exists(case_path):
                os.makedirs(case_path)

            get_input_file_with_schedules_etc(input_file_path, complete_input_file_path_uc,False,int(y))

            with open(complete_input_file_path_uc, "r+", encoding="utf-8") as f:
                    data = json.load(f)
                    data["natural_ventilation_method"] = "air changes per hour"
                    data["natural_ventilation_rate_open_windows"] = 2

            with open(complete_input_file_path_c, "w", encoding="utf-8") as f:
                json.dump(data, f, indent=4)


```
