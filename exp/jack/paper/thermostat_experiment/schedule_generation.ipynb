```python
import pandas as pd
import numpy as np

def random_sample(row):
    return np.random.choice(row)

zones = [
        "bathroom",
        "front_room",
        "kitchen",
        "bedroom_1",
        "bedroom_2"
    ]

schedule_path = "/workspaces/CUBES/cubes/data/schedules/combined.sch"
schedule_master = pd.read_csv(schedule_path, index_col=0)
schedule_master.columns = schedule_master.columns.str.lower()

reps = np.arange(5)
for rep in reps:
    occ_schedule = pd.DataFrame()
    for zone in zones:
        zone_schedules = schedule_master.filter(like=zone)
        zone_schedules = zone_schedules.apply(random_sample, axis=1)
        occ_schedule = pd.concat([occ_schedule, zone_schedules], axis=1)

    occ_schedule.columns = zones
    occ_schedule.to_csv(f"/workspaces/CUBES/exp/jack/paper/zoning_experiment/input/rep{rep}/schedule_rep_{rep}")
```
