### Notebook to create building config files with a different single zone being controlled

```python
import pandas as pd
from cubes.construct.buildingconfig import load_building_config
from cubes.construct.building import Building
```

```python
bc_file_path = "/workspaces/CUBES/exp/jack/paper/thermostat_experiment/input/case0/rep0/baseline.json"
files_dir = "/workspaces/CUBES/exp/jack/paper/thermostat_experiment/files_dir"

BC = load_building_config(bc_file_path, files_dir=files_dir)

def create_buildingconfigs_iters(BC):

    # Change range for many schedule reps
    for n in range(1):
        occ_filename = f"thermostat_exp/schedule_rep_{n}.sch"
        BC.ocupant_schedule_file_name = occ_filename

        # Range for the number of zones in the building
        for m in range(9):

            all_zones = [item for sublist in BC.zone_names for item in sublist]

            BC.controlled_zones = [all_zones[m]]

            BC.gas_pricing_file_name = "csv_gastracker_A_Eastern_England_2022.csv"
            BC.electricity_pricing_file_name = "csv_agile_A_Eastern_England_2022.csv"

            BC.save_to_file(f"/workspaces/CUBES/exp/jack/paper/thermostat_experiment/input/case0/rep{n}/iter{m}.json")
```

```python
create_buildingconfigs_iters(BC)
```
