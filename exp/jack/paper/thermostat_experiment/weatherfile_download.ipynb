```python
import requests

# Define the coordinates for Cambridge
latitude = 52.2053
longitude = 0.1218

# Specify the region name and year
cleaned_region_name = "cambridge"
year = 2023
epw_file_name = f"{cleaned_region_name}_{year}.epw"

# Your Oikolab API key
OIKOLAB_API_KEY = "6f8a9cf2311245a783c97c7275de97b2"

# Make the request to Oikolab API
r = requests.get(
    "https://api.oikolab.com/epw",
    params={'lat': latitude,
            'lon': longitude,
            'year': year},
    headers={'api-key': OIKOLAB_API_KEY},
    timeout=10,
)

# Check the response
if r.status_code == 200:
    with open(epw_file_name, 'wb') as f:
        f.write(r.content)
    print(f"EPW file saved as {epw_file_name}")
else:
    print(f"Failed to retrieve data: {r.status_code}, {r.text}")
```
