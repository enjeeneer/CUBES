```python
import pandas as pd

years = [2022, 2023]

energy_vector = ["csv_agile_A_Eastern_England", "csv_agileoutgoing_A_Eastern_England", "csv_gastracker_A_Eastern_England"]
```

```python
def create_schedule(years, vectors):

    for year in years:
        yearly_data = []
        for vector in vectors:
            df = pd.read_csv(vector+".csv", names=["timestamp", "time", "region_short", "region_full", "price"])

            # Convert 'timestamp' column to datetime
            df['timestamp'] = pd.to_datetime(df['timestamp'])
            df.set_index('timestamp', inplace=True)

            # Filter for the specific year
            df = df[df.index.year == year]

            # Resample to 10-minute intervals and interpolate
            if "gas" in vector:
                df_resampled = df.resample('10min').ffill()
            else:
                df_resampled = df.resample('10min').interpolate()
            df_resampled.reset_index(inplace=True)


            # Forward fill NaN values
            df_resampled[['region_short', 'region_full']] = df_resampled[['region_short', 'region_full']].ffill()

            # Recompute the 'time' column from the 'timestamp'
            df_resampled['time'] = df_resampled['timestamp'].dt.strftime('%H:%M')

            df_resampled.to_csv(f"{vector}_{year}_full.csv", index=False)
            df_resampled_schedule = df_resampled[["timestamp","price"]]
            df_resampled_schedule.to_csv(f"{vector}_{year}.csv", index=False)

    return df_resampled_schedule
```

```python
df = create_schedule(years, energy_vector)
```
