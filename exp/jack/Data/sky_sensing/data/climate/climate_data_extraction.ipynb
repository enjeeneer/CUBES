```python
import pandas as pd
import glob
```

```python
jsonfiles = []
for file in glob.glob("*.json"):
    jsonfiles.append(file)

def convert_json_to_df(jsonfiles):
    for filename in jsonfiles:
        df = pd.read_json(filename)
        data = {'timestamp':[], 'temperature':[]}

        for series_name, series in df.items():

            if series.iloc[0]['state'] == "unavailable":
                data['timestamp'].append(series.iloc[0]['last_changed'])
                data['temperature'].append("")
            else:
                data['timestamp'].append(series.iloc[0]['last_changed'])
                data['temperature'].append(series.iloc[0]['attributes']['current_temperature'])

        data = pd.DataFrame.from_dict(data)
        savename = filename.split(".")[0] + "." + filename.split(".")[1]
        data.to_csv(savename+".csv", index=False)

convert_json_to_df(jsonfiles)
```
