### Script to Invert Occupancy data - door is always closed when the room is occupied and vice versa

```python
import pandas as pd
import pathlib

src=pathlib.Path("/workspaces/CUBES/beizaee_validation/misc/input/occupancy")
dst=pathlib.Path("/workspaces/CUBES/beizaee_validation/misc/input/door")
dst.mkdir(exist_ok=True)

for f in src.glob("*.sch"):
    df=pd.read_csv(f)
    for c in df.columns:
        if pd.api.types.is_numeric_dtype(df[c]):
            df[c]=df[c].apply(lambda x: 0 if x>0 else 1)
    new_name="door_"+f.name
    df.to_csv(dst/new_name,index=False)
```
